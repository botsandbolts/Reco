## Why chronological splits?
We hold one final event per user for *test* and a preceding event for validation. The remaining held event trains the LambdaMART ranker. The early positive ratings alone fit SVD and the two-tower model. This prevents obvious user-level future-item leakage, but does **not** create a global chronological cutoff.
